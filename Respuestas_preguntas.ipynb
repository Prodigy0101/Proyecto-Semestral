¿Cuánto ha aumentado la existencia o exportación de aparatos electrónicos en Chile
en el siglo XXI?

Datos_asociados : ComputadoresCelularesChile.xlsx

Imagen Asociada:  grafico_exportacion_dispositivos

Según los datos del Banco Central, los gastos de exportaciones de computadores han mostrado un aumento considerable durante el pasar de los años (2003 a 2025). Presentando un mínimo de menos de 10 millones USD en 2003 y alcanzando un máximo muy alto de alrededor de 130 millones USD en 2020-2024. Por otra parte, los teléfonos celulares presentan también una gran alza en los gastos de exportación, con un mino un poco superior a los 10 millones de USD en 2003, y un máximo enorme de 250 millones USD en 2012-2016.
En el periodo analizado, el mercado de exportaciones de dispositivos electrónicos en Chile ha demostrado un crecimiento gradual al pasar de los años, siendo los celulares los que mayor valor de exportación tienen.
